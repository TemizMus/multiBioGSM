```python
import pandas as pd

# Excel dosyasını oku
df_1 = pd.read_excel('cancer-DisGeNET.xlsx')


# DataFrame'i görüntüle
print(df_1)

```

```python
birlesik_df_1 = pd.concat([df_1['diseaseName'], df_1['geneSymbol']], axis=1)
birlesik_df_1
```

```python
df_yeni = pd.DataFrame(birlesik_df_1)
df_grouped = df_yeni.groupby('diseaseName')['geneSymbol'].apply(', '.join).reset_index()
df_grouped #geneSymbol
```

```python
df_2 = pd.read_excel('miRtarBase_homoSapian.xlsx')

miRNA = df_2['miRNA']
Target_Gene = df_2['Target Gene']

birlesik_df_2 = pd.concat([df_2['miRNA'], df_2['Target Gene']], axis=1)
birlesik_df_2
```

```python
df_yeni_2 = pd.DataFrame(birlesik_df_2)
df_grouped_2 = df_yeni_2.groupby('miRNA')['Target Gene'].apply(', '.join).reset_index()
```

```python
df_GO_MF = pd.read_excel('GO_MF.xlsx')
df_GO_BP = pd.read_excel('GO_BP.xlsx')
df_GO_CC = pd.read_excel('GO_CC.xlsx')
secilen_df_go = pd.concat([df_GO_CC, df_GO_BP, df_GO_MF], ignore_index=True, axis = 0)
secilen_df_go
```

```python
df_grouped.columns = [0, 1]  # Sütun isimlerini sıfırla
df_grouped_2.columns = [0, 1]  # Sütun isimlerini sıfırla
secilen_df_go.columns = [0, 1]  # Sütun isimlerini sıfırla
secilen_df = pd.concat([df_grouped, df_grouped_2,secilen_df_go], ignore_index=True, axis = 0)
secilen_df
```
